```python
# 6
# Numeric listdə min max elementi tapıb onların hasilini tapın. Bu nəticənin 
#listdəki elementlərin cəmindən müqayisədə True və ya False olduğunu çıxarın. (hazir funksiyalar olmaz)


listt=[]
for i in range(5):
    a=int(input())
    listt.append(a)

minn=listt[0]
maxx=listt[0]
summ=0
for i in range(len(listt)):
    summ+=listt[i]
    if listt[i]>maxx:
        maxx=listt[i]
    elif listt[i]<minn:
        minn=listt[i]

hasil=minn*maxx

if hasil<summ:
    print("False")
else:
    print("True")
```

```python
# 20. Change diagonals
a = [[1,2,3],
     [4,5,6],
     [7,8,9]]
n=len(a)
for i in range(len(a)):
    for j in range(len(a[i])):
        if i==0 or i==n-1:
            a[i][j],a[i][n-1]=a[i][n-1],a[i][j]
print(a)
```
