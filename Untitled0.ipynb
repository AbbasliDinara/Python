```python
import seaborn as sns
import matplotlib.pyplot as plt

df=sns.load_dataset('penguins')
df
```

```python
#Her penguin novune gore ceki paylanmasi ve cinsi de ferlqnedirme
sns.boxplot(data = df, x = "species", y="body_mass_g", hue = "sex")
plt.grid()
plt.show()
```

```python

#Her nov penguin ucun beden ceksinin paylanmasi
sns.histplot(data = df , x = "body_mass_g", hue = "species", kde = True)

#sns.barplot(data = df , x ="day", y = "tip", hue = "sex", errorbar=None)
```

```python

sns.kdeplot(data=df, x="body_mass_g", hue="species", fill=True)
plt.show()
```

```python
#adada yasaayan penguin cekisine gore
sns.barplot(data = df , x ="island", y = "body_mass_g", hue = "sex", errorbar=None)
#sns.barplot(data = df , x ="day", y = "tip", hue = "sex", errorbar=None)
```

```python
#her adada yasayan penguin nov sayi
sns.countplot(data=df, x="island", hue="species")
plt.show()
```

```python
#renge gore novu ayir, ceki sizela dimsik olcusunun qanaddan asililigi

sns.scatterplot(data=df, x="flipper_length_mm", y="bill_length_mm",size="body_mass_g",hue="species",legend=False)
plt.show()
```

```python
# her nov penguin ucun length paylanma novu ile
sns.violinplot(data=df, x="species", y="bill_length_mm", hue="sex")
plt.show()
```

```python
corr = df[['bill_length_mm','bill_depth_mm','flipper_length_mm']].corr()
sns.heatmap(corr, annot=True, cmap="mako_r")
plt.show()

#sns.heatmap(df.corr(numeric_only = True),annot=True , cmap="mako_r")

```
