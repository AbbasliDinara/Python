```python
'''
Outlier analizi

Ən azı iki ədədi dəyişən üçün outlier-ları:

boxplot ilə göstərin,

nəticəni qısa şəkildə şərh edin.

'''
# Outlier hesablamaları
q1 = np.quantile(df["Final_Score"], 0.25)
q3 = np.quantile(df["Final_Score"], 0.75)
iqr = q3 - q1 
minn = q1 - 1.5* iqr
maxx = q3 + 1.5 *iqr
fig, ax = plt.subplots(2, 1, figsize=(8,8))
ax[0].hist(df["Final_Score"])
ax[0].vlines(x = minn , ymin = 0 , ymax = 350, color="red")
ax[0].vlines(x = maxx , ymin = 0 , ymax = 350, color="red")

sns.boxplot(data = df , x = "Final_Score")

##kenara cixma yoxdur demdeli butun valular IQR metoduna uyugundur 

```

```python
 # Outlier hesablamaları
q1 = np.quantile(df["Midterm_Score"], 0.25)
q3 = np.quantile(df["Midterm_Score"], 0.75)
iqr = q3 - q1 
minn = q1 - 1.5* iqr
maxx = q3 + 1.5 *iqr
fig, ax = plt.subplots(2, 1, figsize=(8,8))
ax[0].hist(df["Midterm_Score"])
ax[0].vlines(x = minn , ymin = 0 , ymax = 350, color="red")
ax[0].vlines(x = maxx , ymin = 0 , ymax = 350, color="red")

sns.boxplot(data = df , x = "Midterm_Score",color="green")

##kenara cixma yoxdur demdeli butun valular IQR metoduna uyugundur 
```
