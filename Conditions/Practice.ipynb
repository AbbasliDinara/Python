```python
# 5
# İstifadəçidən x ve y ededleri alın, daxil edilən ədədin koordinat müstəvisində neçənci rübdə olduğunu tapan kod yazın. (x = 4, y = -6 olarsa 4cu rub) 
# x, y oxlari uzerinde, koordinat baslangicinda olmaq kimi sertleri de nezere alin
x=int(input())
y=int(input())
if x==0 and y==0:
    print("koordinant baslangicindadir")
elif x==0 and y!=0:
    print("y oxu uzerindedir")
elif y==0 and x!=0:
    print("x oxu uzerindedir")
elif x>0:
    if y>0:
        print("I rub")
    else:
        print("IV rub")
elif x<0:
    if y>0:
        print("II rub")
    else:
        print("III rub")
```
