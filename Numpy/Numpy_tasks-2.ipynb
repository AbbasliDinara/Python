```python
import numpy as np
a = np.random.randn(3, 3)
b = np.where(a>0, '+', '-')
b
```

```python
# 12
# loopdan istifade ederek a arrayine esasen b arrayini ozunuz yaradin

b_new=np.empty((3, 3), dtype='str')
b_new

for i in range(len(a)):
    for j in range(len(a[i])):
        if a[i][j]>0:
            b_new[i][j]='+'
        else:
            b_new[i][j]='-'
b_new
```
