```python
#17 list kimi

   
arr20= np.array([0, 10, 20, 40, 60])
arr21 = np.array([0, 40])
result=[]
for i in range(len(arr20)):
    if arr20[i] in arr21:
        result.append(True)
    else:
        result.append(False)
print(result)

#17 array kimi
arr20 = np.array([0, 10, 20, 40, 60])
arr21 = np.array([0, 40])

print(np.isin(arr20, arr21)) ##in1t
```

```python
#18 list kimi
arr22= np.array([0, 10, 20, 40,60])
arr23 = np.array([10,30,40])
result=[]
for i in range(len(arr23)):
    if arr23[i] in arr22:
        result.append(int(arr23[i]))

print(result)


#18 array kimi
arr22 = np.array([0, 10, 20, 40,60])
arr23 = np.array([10,30,40])
com_el = np.intersect1d(arr22, arr23)
print(com_el)
```
