```python
# 1
# Write a Python function to multiply all the numbers in a list.
# Sample List : [8, 2, 3, -1, 7]

def multiply(listt):
    result=1

    for i in range(len(listt)):
        result=listt[i]*result
    return result

print(multiply([8, 2, 3, -1, 7]))



```

```python
# 2
# Write a Python program to reverse a string using while loop
# Sample String : "1234abcd"
# Expected Output : "dcba4321"

def reverse(string):
    result=""
    i=len(string)-1
    
    while i>=0:
        result+=string[i]
        i-=1
    return result
print(reverse("1234abcd"))
```

```python
# 3
# Find count of max
# [3,4,5,71,65,71,44,71,62]

def max_count(listt):
    count=0
    maxx=listt[0]
    for i in range(len(listt)):
        if maxx<listt[i]:
            maxx=listt[i]
            count=1
        elif listt[i]==maxx:
            count+=1
    return count
print(max_count([3,4,5,71,65,71,44,71,62]))
        
```

```python
# 4
# x ** y tapan funksiya yaradin, daxiline loop istifade olunsun
# birbasa x ustu y yazmayin
def power(x,y):
    result=1
    for i in range(y):
        result=x*result
    return result
print(power(2,3))
```

```python
# 5
# Pig Latin Converter:    
# Eger soz saitle baslayirsa sozun sonuna 'way' artirilir. 
# Samitle baslayirsa ilk herf sona gelir ve sonuna 'ay' artirilir
# meselen:   
# apple -- appleway
# banana -- ananabay

def converter(string):
    saitler="aeiou"
    samitler="bcdfghjklmnpqrstvwxyz"
    if string[0] in saitler:
        return string + "way"
    if string[0] not in saitler:
        return string[1:]+string[0]+"ay"

print(converter("apple"))
print(converter("banana"))
        
        
```

```python
# 6
# Unique elemetlerin sayini tapan funksiya yaradin --- {element1 : sayi, element2 : sayi}
# input : [4,1,2,2,3,1,1]
# output : {4: 1, 1: 3, 2: 2, 3: 1} 



def find_unique(listt):
    
    result=dict()
    for i in range(len(listt)):
        if listt[i] in result:
            result[listt[i]]+=1
        else:
            result[listt[i]]=1
    return result

print(find_unique([4,1,2,2,3,1,1]))
        
    
```

```python
# 7
# Recursive function
# verilmis edede qeder butun ededlerin cemini tapan recursive funct yaradin
# input: 4
# output: 6 (1+2+3)

def summ(n):
    if n == 0:
        return 0
    return n-1 + summ(n-1)

print(summ(4))  

```

```python
# 8
# 5**4 hesablamaq ucun recursive funct yaradin

def power(x,y):
    result=1
    for i in range(y):
        result=x*result
    return result
print(power(5,4))
```

```python
# 9
# factorial(3) = 3 * factorial(2) = 3 * 2 * factorial(1)= 3 * 2 * 1 * factorial(0) = 3 * 2 * 1 * 1 = 6
# Factorial i hesablayan funksiya yaradin


def factorial(n):
    if n == 0 or n==1:
        return 1
    return n * factorial(n-1) 
    
print(factorial(3))  
```

```python
# 11
# N ədədi daxil edilə bilən bir minMax adlıfunksiya yaradın.
# Funksiya n dəfə yeni input olaraq ədəd almalıdır user’dən.
# Funksiya bu ədədlər arasında minimumu maksimum ədəd dərəcəsindən yüksəldərək return versin.
# Misal: minMax(5)
# Input: 2 3 4 5 6
# Output: 64


def minMax(n):
    a=int(input())
    maxx=a
    minn=a
    for i in range(n-1):
        a=int(input())

        if a>maxx:
            maxx=a
        if a<minn:
            minn=a
    return minn**maxx
    
print(minMax(5))
        
```

```python
# 12
# String düz və tərsinə oxunduqda eyni olursa, polindrom
# string sayılır. Checker adlı funksiya yaradın və daxil edilən
# stringin polindrom olub-olmamasını yoxlasın uyğun olaraq
# boolean tipində true vəya false return etsin.

def checker(string):
    string = string.lower()
    return string == string[::-1]

    


print(checker("abssba"))   
print(checker("hello"))  
print(checker("Dinara"))  
```

```python
# 13
# Fibonacci adlı n ədədi daxil ediləbilən bir funksiya yaradın.
# Bu funksiya 0-dan başlayaraq n ədəd fibonacci ədədlərini bir
# listin içinə yığsın. Yekunda return olaraq bu listi versin.

def fibonacci(n):

    result=[]
    first=0
    second=1

    for i in range(n):
        temp=second
        second=first+second
        first=temp
        result.append(first)
    return result


print(fibonacci(7))
```

```python
# 14
# Birinci sinif şagirdlərinə əgər çəkisi 30 kq-dan az olarsa, əlavə olaraq bir stəkan süd və tort verilir.
# Süd stəkanının tutumu 200 ml, süd paketi isə - 0,9 litrdir. 
# Hər gün tələb olunan əlavə süd paketlərinin və tortların sayını təyin edin.
# Evvelce n sayda birinci sinif sagirdlerinin cekilerini daxil edin, sonra cekisinin 30kq dan az olanlari sayin
# Məsələn: 20 usaqdan cəkisi 30 kqdan aşaği olan 8 nefer var.
# Deməli, 8 tort və 2 paket süd tələb olunur.

def extra(kilos):

    count = 0
    
    for kg in kilos:
        if kg < 30:
            count += 1
    totalmilk = count * 200
    
    paket = totalmilk / 900
    import math
    paket = math.ceil(paket) ## yuvarla
    
  
    tort = count
    
    return tort, paket


student_wei = [28, 32, 29, 12, 27, 31, 26, 33, 56, 30, 28, 34, 53, 29, 98, 36, 25, 17, 28, 30]

tort, paket = extra(student_wei)
print("tort sayı:", tort)
print("paket sayı:", paket)

```

```python
# 29
# def, filter, lambda funksiyalarindan (3u birlikde) istifade ederek uzunlugu 5den boyuk olan sozleri qaytarin
soz=["apple", "banana", "orange", "grapes", "watermelon"]
# output : ['banana', 'orange', 'grapes', 'watermelon']


#filter(function, iterable)

def long_word(listt):
    return list(filter(lambda x: len(x) > 5, listt))

# Funksiyanı çağırırıq
result = long_word(soz)
print(result)
```

```python
# 30
# filter ve lambda funksiyalarindan (2si birlikde) istifade ederek samitleri qaytaran funksiya yaradin
soz= ['a', 'b', 'c', 'd', 'e', 'f']
# output : ['b', 'c', 'd', 'f']
samitler="bcdfghjklmnpqrstvwxyz"
def find_samit(listt):
    return list(filter(lambda x:  x in samitler, listt))

# Funksiyanı çağırırıq
result = find_samit(soz)
print(result)
```

```python
# 31
# lambda, map, filter funksiyalarindan (3u birlikde) istifade ederek (1, 11) araliginda cut ededlerin kvadratini tapin
# output : [4, 16, 36, 64, 100]



def find_odd(listt):
    return list(map(lambda x: x**2 ,filter(lambda x:  x%2==0, listt)))

# Funksiyanı çağırırıq
numbers = range(1, 11)
result = find_odd(numbers)
print(result)
```

```python
# 32
# lambda, map, filter funksiyalarindan (3u birlikde) istifade ederek icinde 'a' herfi olan sozlerin uzunlugunu qaytarin
# input : ["apple", "banana", "orange", "kiwi"]
# output : [5, 6, 6]


def find_len(listt):
    return list(map(lambda x: len(x) ,filter(lambda x:  'a' in x, listt)))

# Funksiyanı çağırırıq
numbers=["apple", "banana", "orange", "kiwi"]
result = find_len(numbers)
print(result)
```
