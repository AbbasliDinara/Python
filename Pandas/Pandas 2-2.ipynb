# JOINS

```python
df = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3', 'K4', 'K5'],
                   'A': ['A0', 'A1', 'A2', 'A3', 'A4', 'A5']})
```

```python
df
```

```python
other = pd.DataFrame({'key': ['K0', 'K1', 'K2'],
                      'B': ['B0', 'B1', 'B2']})
```

```python
other
```

```python
df.merge(other, how = "right", on = "key")
```

```python
df.merge(other, how = "left", on = "key")
```

```python
df.merge(other, how = "outer", on = "key")
```

```python
df.merge(other, how = "inner", on = "key")
```
