```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("titanic.csv")
```

```python
df
```

```python
#2.Qadın sərnişinlərin ortalama yaşları neçədir?
df[df["Sex"]=="female"].mean(numeric_only=True)[["Age"]]
```

```python
#3."PassengerId", "Name" və "Ticket" sütunlarını silin və yenidən dataseti çağırdıqda o sütunlar görünməsin

df.drop("PassengerId" , axis = 1, inplace=True) # ve ya df.drop("C1", axis = 1, inplace = True)
df.drop("Name" , axis = 1, inplace=True)
df.drop("Ticket" , axis = 1, inplace=True)
```

```python
df
```

```python
#4.Gediş haqqı üçün 7000 və ondan az ödəmə edən müştərilərin sayın gətirin (məlumatlar gəlməsin sadəcə say gəlsin)
df[df["Fare"]<=7000].count()[["Fare"]]
```

```python
#5.Həyatda qalan sərnişinlər toplam sərnişinlərin neçə faizidir?
percentage=df[df["Survived"]==1].count()[["Survived"]]/df.count()[["Survived"]] *100
print(round(percentage,2))
```

```python
#6.ən yaşlı və ən gənc sərnişinlərin yaşı neçədir?
young=df["Age"].min()
old=df["Age"].max()
old
```
