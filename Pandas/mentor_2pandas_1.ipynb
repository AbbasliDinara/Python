```python
df[" Income "] = (
    df[" Income "]
    .astype(str)
    .str.strip()
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.replace(".00", "", regex=False)
    .astype(int)
) ## errorlu
```

```python
df[" Income "] = (df[" Income "].astype(str)                       # nan-lar string oldu  
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.replace(".00", "", regex=False)
)
df[" Income "] = pd.to_numeric(df[" Income "], errors="coerce").astype("Int64") ##inte cevirdi

df
```
